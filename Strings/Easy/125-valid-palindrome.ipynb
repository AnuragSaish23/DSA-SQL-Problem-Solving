# Valid Palindrome (LeetCode #125)

## Problem Description

A phrase is a **palindrome** if, after converting all uppercase letters into lowercase letters and removing all non-alphanumeric characters, it reads the same forward and backward. Alphanumeric characters include letters and numbers.

Given a string `s`, return `true` if it is a **palindrome**, or `false` otherwise.

### Examples

**Example 1:**
```
Input: s = "A man, a plan, a canal: Panama"
Output: true
Explanation: "amanaplanacanalpanama" is a palindrome.
```

**Example 2:**
```
Input: s = "race a car"
Output: false
Explanation: "raceacar" is not a palindrome.
```

**Example 3:**
```
Input: s = " "
Output: true
Explanation: s is an empty string "" after removing non-alphanumeric characters.
Since an empty string reads the same forward and backward, it is a palindrome.
```

### Constraints
- `1 <= s.length <= 2 * 10^5`
- `s` consists only of printable ASCII characters.

## Approach

**Two Pointer Technique**

1. Use two pointers: `left` starting from the beginning, `right` starting from the end
2. Skip non-alphanumeric characters by moving pointers inward
3. Compare characters (case-insensitive) at both pointers
4. If they don't match, return False
5. If we traverse the entire string without mismatches, return True

````python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left, right = 0, len(s) - 1
        
        while left < right:
            # Skip non-alphanumeric characters from left
            while left < right and not s[left].isalnum():
                left += 1
            
            # Skip non-alphanumeric characters from right
            while left < right and not s[right].isalnum():
                right -= 1
            
            # Compare characters (case-insensitive)
            if s[left].lower() != s[right].lower():
                return False
            
            left += 1
            right -= 1
        
        return True
````

````python
# Test the solution
sol = Solution()

# Test Case 1
print(sol.isPalindrome("A man, a plan, a canal: Panama"))  # Expected: True

# Test Case 2
print(sol.isPalindrome("race a car"))  # Expected: False

# Test Case 3
print(sol.isPalindrome(" "))  # Expected: True
````

## Complexity Analysis

- **Time Complexity:** O(n) - We traverse the string at most once.
- **Space Complexity:** O(1) - Only constant extra space is used for the two pointers.
