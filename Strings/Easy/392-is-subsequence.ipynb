# Is Subsequence (LeetCode #392)

## Problem Description

Given two strings `s` and `t`, return `true` if `s` is a **subsequence** of `t`, or `false` otherwise.

A **subsequence** of a string is a new string that is formed from the original string by deleting some (can be none) of the characters without disturbing the relative positions of the remaining characters. (i.e., `"ace"` is a subsequence of `"abcde"` while `"aec"` is not).

### Examples

**Example 1:**
```
Input: s = "abc", t = "ahbgdc"
Output: true
```

**Example 2:**
```
Input: s = "axc", t = "ahbgdc"
Output: false
```

### Constraints
- `0 <= s.length <= 100`
- `0 <= t.length <= 10^4`
- `s` and `t` consist only of lowercase English letters.

**Follow up:** Suppose there are lots of incoming `s`, say `s1, s2, ..., sk` where `k >= 10^9`, and you want to check one by one to see if `t` has its subsequence. In this scenario, how would you change your code?

## Approach

### Two Pointer - Optimal

**Key Insight:** We need to check if all characters of `s` appear in `t` in the same order. We can use two pointers - one for each string.

**Algorithm:**
1. Initialize two pointers:
   - `i` pointing to the start of `s`
   - `j` pointing to the start of `t`

2. Iterate through `t` with pointer `j`:
   - If `s[i] == t[j]`, we found a match, move `i` forward
   - Always move `j` forward

3. If `i` reaches the end of `s`, all characters were matched in order → return `True`

**Why this works:**
- We're greedily matching characters in `s` as early as possible in `t`.
- This greedy approach is optimal because matching earlier never hurts.

````python
# Python Solution
class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        # Two pointer approach
        i = 0  # Pointer for s
        j = 0  # Pointer for t
        
        while i < len(s) and j < len(t):
            # If characters match, move pointer in s
            if s[i] == t[j]:
                i += 1
            # Always move pointer in t
            j += 1
        
        # If we've matched all characters in s, it's a subsequence
        return i == len(s)


# Test the solution
sol = Solution()

# Test case 1
print(f"Test 1: {sol.isSubsequence('abc', 'ahbgdc')}")  # Expected: True

# Test case 2
print(f"Test 2: {sol.isSubsequence('axc', 'ahbgdc')}")  # Expected: False

# Test case 3 - Empty s
print(f"Test 3: {sol.isSubsequence('', 'ahbgdc')}")  # Expected: True

# Test case 4 - Equal strings
print(f"Test 4: {sol.isSubsequence('abc', 'abc')}")  # Expected: True
````

## Complexity Analysis

- **Time Complexity:** O(n)
  - Where n is the length of string `t`.
  - We traverse `t` once in the worst case.

- **Space Complexity:** O(1)
  - We only use a constant amount of extra space for two pointers.
