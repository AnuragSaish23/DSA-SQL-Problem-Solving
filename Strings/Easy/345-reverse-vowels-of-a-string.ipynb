# Reverse Vowels of a String (LeetCode #345)

## Problem Description

Given a string `s`, reverse only all the vowels in the string and return it.

The vowels are `'a'`, `'e'`, `'i'`, `'o'`, and `'u'`, and they can appear in both lower and upper cases, more than once.

### Examples

**Example 1:**
```
Input: s = "hello"
Output: "holle"
```

**Example 2:**
```
Input: s = "leetcode"
Output: "leotcede"
```

### Constraints
- `1 <= s.length <= 3 * 10^5`
- `s` consist of **printable ASCII** characters.

## Approach

**Two Pointer Technique**

1. Use two pointers: `left` at the beginning, `right` at the end
2. Move `left` forward until we find a vowel
3. Move `right` backward until we find a vowel
4. Swap the vowels at both positions
5. Continue until the pointers meet

This efficiently reverses only the vowels while keeping other characters in place.

````python
class Solution:
    def reverseVowels(self, s: str) -> str:
        vowels = set('aeiouAEIOU')
        s_list = list(s)  # Convert to list for mutability
        
        left, right = 0, len(s) - 1
        
        while left < right:
            # Find vowel from left
            while left < right and s_list[left] not in vowels:
                left += 1
            
            # Find vowel from right
            while left < right and s_list[right] not in vowels:
                right -= 1
            
            # Swap vowels
            s_list[left], s_list[right] = s_list[right], s_list[left]
            
            left += 1
            right -= 1
        
        return ''.join(s_list)
````

````python
# Test the solution
sol = Solution()

# Test Case 1
print(sol.reverseVowels("hello"))  # Expected: "holle"

# Test Case 2
print(sol.reverseVowels("leetcode"))  # Expected: "leotcede"

# Test Case 3
print(sol.reverseVowels("aA"))  # Expected: "Aa"
````

## Complexity Analysis

- **Time Complexity:** O(n) - We traverse the string at most once.
- **Space Complexity:** O(n) - We convert the string to a list for in-place modifications. The vowel set is O(1) as it contains a fixed number of characters.
