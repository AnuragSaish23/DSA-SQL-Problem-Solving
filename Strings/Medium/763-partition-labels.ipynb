# Partition Labels (LeetCode #763)

## Problem Description

You are given a string `s`. We want to partition the string into as many parts as possible so that each letter appears in at most one part.

Note that the partition is done so that after concatenating all the parts in order, the resultant string should be `s`.

Return a list of integers representing the size of these parts.

### Example 1:
```
Input: s = "ababcbacadefegdehijhklij"
Output: [9,7,8]
Explanation:
The partition is "ababcbaca", "defegde", "hijhklij".
This is a partition so that each letter appears in at most one part.
A partition like "ababcbacadefegde", "hijhklij" is incorrect, because it splits s into less parts.
```

### Example 2:
```
Input: s = "eccbbbbdec"
Output: [10]
```

### Constraints:
- `1 <= s.length <= 500`
- `s` consists of lowercase English letters.

## Approach 1: Brute Force

**Intuition:** For each starting position, find the minimum valid partition by ensuring all characters in that partition don't appear later in the string.

**Algorithm:**
1. Start from index 0
2. For current partition, find the last occurrence of each character
3. Extend the partition until we've included all last occurrences
4. Record partition size and move to next partition

````python
# Brute Force Solution
class Solution:
    def partitionLabels(self, s: str) -> list[int]:
        result = []
        start = 0
        
        while start < len(s):
            # Find the end of current partition
            end = start
            i = start
            
            while i <= end:
                # Find last occurrence of current character
                last_occur = s.rfind(s[i])
                end = max(end, last_occur)
                i += 1
            
            result.append(end - start + 1)
            start = end + 1
        
        return result
````

**Complexity Analysis (Brute Force):**
- Time: O(n²) - For each character, rfind takes O(n)
- Space: O(1) - Only storing result (not counting output)

---
## Approach 2: Optimal (Greedy with Last Occurrence Map)

**Intuition:** Pre-compute the last occurrence of each character. Then use two pointers - one for partition start and one for partition end. Extend the end whenever we encounter a character whose last occurrence is beyond current end.

**Algorithm:**
1. Build a map of last occurrence index for each character
2. Initialize start = 0, end = 0
3. For each character, update end = max(end, last[char])
4. When current index equals end, we've found a complete partition
5. Record partition size and start new partition

````python
# Optimal Solution - Greedy with Last Occurrence Map
class Solution:
    def partitionLabels(self, s: str) -> list[int]:
        # Build last occurrence map
        last_occurrence = {}
        for i, char in enumerate(s):
            last_occurrence[char] = i
        
        result = []
        start = 0
        end = 0
        
        for i, char in enumerate(s):
            # Extend partition end to include last occurrence of current char
            end = max(end, last_occurrence[char])
            
            # If we've reached the end of current partition
            if i == end:
                result.append(end - start + 1)
                start = i + 1
        
        return result
````

**Complexity Analysis (Optimal):**
- Time: O(n) - Two passes through the string
- Space: O(1) - At most 26 characters in the map (constant)

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n²) | O(1) |
| Optimal (Greedy) | O(n) | O(1) |

**Key Insight:** By pre-computing last occurrences, we can determine partition boundaries in a single pass using the greedy approach.
