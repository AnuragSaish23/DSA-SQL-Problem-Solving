# Merge Sorted Array (LeetCode #88)

## Problem Description

You are given two integer arrays `nums1` and `nums2`, sorted in **non-decreasing order**, and two integers `m` and `n`, representing the number of elements in `nums1` and `nums2` respectively.

**Merge** `nums1` and `nums2` into a single array sorted in **non-decreasing order**.

The final sorted array should not be returned by the function, but instead be stored inside the array `nums1`. To accommodate this, `nums1` has a length of `m + n`, where the first `m` elements denote the elements that should be merged, and the last `n` elements are set to `0` and should be ignored. `nums2` has a length of `n`.

### Examples

**Example 1:**
```
Input: nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3
Output: [1,2,2,3,5,6]
Explanation: The arrays we are merging are [1,2,3] and [2,5,6].
The result of the merge is [1,2,2,3,5,6].
```

**Example 2:**
```
Input: nums1 = [1], m = 1, nums2 = [], n = 0
Output: [1]
```

**Example 3:**
```
Input: nums1 = [0], m = 0, nums2 = [1], n = 1
Output: [1]
```

### Constraints
- `nums1.length == m + n`
- `nums2.length == n`
- `0 <= m, n <= 200`
- `1 <= m + n <= 200`
- `-10^9 <= nums1[i], nums2[j] <= 10^9`

**Follow up:** Can you come up with an algorithm that runs in `O(m + n)` time?

## Approach

### Two Pointer (Reverse) - Optimal

**Key Insight:** Instead of merging from the beginning (which would require shifting elements), we merge from the **end** of `nums1`. This way, we can fill the array in-place without any extra space.

**Algorithm:**
1. Start three pointers:
   - `p1` at the last valid element of `nums1` (index `m - 1`)
   - `p2` at the last element of `nums2` (index `n - 1`)
   - `p` at the end of `nums1` (index `m + n - 1`)

2. Compare elements from the end of both arrays:
   - Place the larger element at position `p`
   - Move the respective pointer backward

3. If `nums2` still has remaining elements, copy them to `nums1`.

**Why this works:**
- The extra space at the end of `nums1` is exactly where we start filling.
- Since we fill from largest to smallest, we never overwrite elements we still need.

````python
# Python Solution
from typing import List

class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        # Start from the end of both arrays
        p1 = m - 1  # Pointer for nums1 valid elements
        p2 = n - 1  # Pointer for nums2
        p = m + n - 1  # Pointer for placement position
        
        # Merge from the end
        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                nums1[p] = nums1[p1]
                p1 -= 1
            else:
                nums1[p] = nums2[p2]
                p2 -= 1
            p -= 1
        
        # If nums2 has remaining elements, copy them
        # (No need to copy remaining nums1 elements - they're already in place)
        while p2 >= 0:
            nums1[p] = nums2[p2]
            p2 -= 1
            p -= 1


# Test the solution
sol = Solution()

# Test case 1
nums1 = [1, 2, 3, 0, 0, 0]
sol.merge(nums1, 3, [2, 5, 6], 3)
print(f"Test 1: {nums1}")  # Expected: [1, 2, 2, 3, 5, 6]

# Test case 2
nums1 = [1]
sol.merge(nums1, 1, [], 0)
print(f"Test 2: {nums1}")  # Expected: [1]

# Test case 3
nums1 = [0]
sol.merge(nums1, 0, [1], 1)
print(f"Test 3: {nums1}")  # Expected: [1]
````

## Complexity Analysis

- **Time Complexity:** O(m + n)
  - We traverse each element of both arrays exactly once.

- **Space Complexity:** O(1)
  - We only use a constant amount of extra space for pointers.
  - The merge is done in-place.
