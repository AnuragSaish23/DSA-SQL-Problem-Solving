# Two Sum (LeetCode #1)

## Problem Description

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

You may assume that each input would have **exactly one solution**, and you may not use the same element twice.

You can return the answer in any order.

### Examples

**Example 1:**
```
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
```

**Example 2:**
```
Input: nums = [3,2,4], target = 6
Output: [1,2]
```

**Example 3:**
```
Input: nums = [3,3], target = 6
Output: [0,1]
```

### Constraints
- `2 <= nums.length <= 10^4`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`
- Only one valid answer exists.

## Approach

**HashMap (Optimal Solution)**

The key insight is that for each number `num` in the array, we need to find if `target - num` exists in the array.

1. Create a HashMap to store each number and its index as we iterate
2. For each number, check if its complement (`target - num`) already exists in the HashMap
3. If found, return the indices; otherwise, add the current number to the HashMap

This approach allows us to find the pair in a single pass through the array.

````python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # HashMap to store number -> index mapping
        num_map = {}
        
        for i, num in enumerate(nums):
            complement = target - num
            
            # Check if complement exists in our map
            if complement in num_map:
                return [num_map[complement], i]
            
            # Store current number with its index
            num_map[num] = i
        
        return []  # No solution found
````

````python
# Test the solution
sol = Solution()

# Test Case 1
print(sol.twoSum([2, 7, 11, 15], 9))  # Expected: [0, 1]

# Test Case 2
print(sol.twoSum([3, 2, 4], 6))  # Expected: [1, 2]

# Test Case 3
print(sol.twoSum([3, 3], 6))  # Expected: [0, 1]
````

## Complexity Analysis

- **Time Complexity:** O(n) - We traverse the list only once, and HashMap lookups are O(1) on average.
- **Space Complexity:** O(n) - In the worst case, we store all n elements in the HashMap.
