# Move Zeroes (LeetCode #283)

## Problem Description

Given an integer array `nums`, move all `0`'s to the end of it while maintaining the relative order of the non-zero elements.

**Note:** You must do this in-place without making a copy of the array.

### Examples

**Example 1:**
```
Input: nums = [0,1,0,3,12]
Output: [1,3,12,0,0]
```

**Example 2:**
```
Input: nums = [0]
Output: [0]
```

### Constraints
- `1 <= nums.length <= 10^4`
- `-2^31 <= nums[i] <= 2^31 - 1`

## Approach

**Two Pointer Technique (Optimal)**

1. Use a pointer `left` to track where the next non-zero element should be placed
2. Iterate through the array with pointer `right`
3. When we find a non-zero element, swap it with the element at `left` position
4. Increment `left` after each swap

This approach efficiently moves all non-zero elements to the front while pushing zeros to the end.

````python
from typing import List

class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        left = 0  # Position for next non-zero element
        
        for right in range(len(nums)):
            if nums[right] != 0:
                # Swap non-zero element to the left position
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
````

````python
# Test the solution
sol = Solution()

# Test Case 1
nums1 = [0, 1, 0, 3, 12]
sol.moveZeroes(nums1)
print(nums1)  # Expected: [1, 3, 12, 0, 0]

# Test Case 2
nums2 = [0]
sol.moveZeroes(nums2)
print(nums2)  # Expected: [0]

# Test Case 3
nums3 = [1, 0, 0, 0, 2]
sol.moveZeroes(nums3)
print(nums3)  # Expected: [1, 2, 0, 0, 0]
````

## Complexity Analysis

- **Time Complexity:** O(n) - We traverse the array only once.
- **Space Complexity:** O(1) - We modify the array in-place without using extra space.
