# Contains Duplicate II (LeetCode #219)

## Problem Description

Given an integer array `nums` and an integer `k`, return `true` if there are two **distinct indices** `i` and `j` in the array such that `nums[i] == nums[j]` and `abs(i - j) <= k`.

### Example 1:
```
Input: nums = [1,2,3,1], k = 3
Output: true
```

### Example 2:
```
Input: nums = [1,0,1,1], k = 1
Output: true
```

### Example 3:
```
Input: nums = [1,2,3,1,2,3], k = 2
Output: false
```

### Constraints:
- `1 <= nums.length <= 10^5`
- `-10^9 <= nums[i] <= 10^9`
- `0 <= k <= 10^5`

## Approach 1: Brute Force

**Intuition:** For each element, check all elements within a distance of k to see if any duplicate exists.

**Algorithm:**
1. For each index i
2. Check all indices j from i+1 to min(i+k, n-1)
3. If nums[i] == nums[j], return True
4. If no duplicates found, return False

````python
# Brute Force Solution
class Solution:
    def containsNearbyDuplicate(self, nums: list[int], k: int) -> bool:
        n = len(nums)
        
        for i in range(n):
            for j in range(i + 1, min(i + k + 1, n)):
                if nums[i] == nums[j]:
                    return True
        
        return False
````

**Complexity Analysis (Brute Force):**
- Time: O(n × k) - For each element, check up to k neighbors
- Space: O(1) - No extra space used

---
## Approach 2: Better (Hash Map)

**Intuition:** Store the last seen index of each element in a hash map. When we encounter a duplicate, check if the distance between current and last seen index is <= k.

**Algorithm:**
1. Create a hash map to store value -> last seen index
2. For each element, check if it exists in the map and distance <= k
3. Update the map with the current index

````python
# Better Solution - Hash Map
class Solution:
    def containsNearbyDuplicate(self, nums: list[int], k: int) -> bool:
        last_seen = {}  # value -> last index
        
        for i, num in enumerate(nums):
            if num in last_seen and i - last_seen[num] <= k:
                return True
            last_seen[num] = i
        
        return False
````

**Complexity Analysis (Better):**
- Time: O(n) - Single pass through the array
- Space: O(n) - Hash map stores up to n entries

---
## Approach 3: Optimal (Sliding Window with Hash Set)

**Intuition:** Maintain a sliding window of size k using a hash set. The set only contains elements within the current window. If we find a duplicate in the set, we return True. This limits space to O(k) instead of O(n).

**Algorithm:**
1. Use a set to maintain a window of at most k elements
2. For each element, check if it's already in the set (duplicate within range k)
3. Add the current element to the set
4. If window size exceeds k, remove the oldest element

````python
# Optimal Solution - Sliding Window with Hash Set
class Solution:
    def containsNearbyDuplicate(self, nums: list[int], k: int) -> bool:
        window = set()
        
        for i in range(len(nums)):
            # Check if duplicate exists in current window
            if nums[i] in window:
                return True
            
            # Add current element to window
            window.add(nums[i])
            
            # Maintain window size of k
            if len(window) > k:
                window.remove(nums[i - k])
        
        return False
````

**Complexity Analysis (Optimal):**
- Time: O(n) - Single pass through the array
- Space: O(k) - Set stores at most k elements

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n × k) | O(1) |
| Better (Hash Map) | O(n) | O(n) |
| Optimal (Sliding Window Set) | O(n) | O(k) |

**Key Insight:** The sliding window set approach combines the speed of hashing with bounded space by only keeping elements within the k-distance window.
