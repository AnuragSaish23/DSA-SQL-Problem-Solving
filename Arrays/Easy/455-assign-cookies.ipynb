# Assign Cookies (LeetCode #455)

## Problem Description

Assume you are an awesome parent and want to give your children some cookies. But, you should give each child at most one cookie.

Each child `i` has a greed factor `g[i]`, which is the minimum size of a cookie that the child will be content with; and each cookie `j` has a size `s[j]`. If `s[j] >= g[i]`, we can assign the cookie `j` to the child `i`, and the child `i` will be content.

Your goal is to maximize the number of your content children and output the maximum number.

**Example 1:**
```
Input: g = [1,2,3], s = [1,1]
Output: 1
Explanation: You have 3 children and 2 cookies. The greed factors of 3 children are 1, 2, 3. 
And even though you have 2 cookies, since their size is both 1, you could only make the child whose greed factor is 1 content.
You need to output 1.
```

**Example 2:**
```
Input: g = [1,2], s = [1,2,3]
Output: 2
Explanation: You have 2 children and 3 cookies. The greed factors of 2 children are 1, 2. 
You have 3 cookies and their sizes are big enough to gratify all of the children, 
You need to output 2.
```

**Constraints:**
- `1 <= g.length <= 3 * 10^4`
- `0 <= s.length <= 3 * 10^4`
- `1 <= g[i], s[j] <= 2^31 - 1`

## Approach 1: Brute Force

**Intuition:** For each child, try to find the smallest cookie that satisfies them. Mark used cookies.

**Algorithm:**
1. For each child's greed factor, find the smallest available cookie that satisfies them
2. Mark the cookie as used
3. Count satisfied children

````python
# Brute Force Solution
class Solution:
    def findContentChildren(self, g: list[int], s: list[int]) -> int:
        if not s:
            return 0
        
        used = [False] * len(s)
        count = 0
        
        # Sort greed factors to satisfy less greedy children first
        g.sort()
        
        for greed in g:
            min_cookie_idx = -1
            min_cookie_size = float('inf')
            
            # Find smallest unused cookie that satisfies this child
            for j, cookie in enumerate(s):
                if not used[j] and cookie >= greed and cookie < min_cookie_size:
                    min_cookie_idx = j
                    min_cookie_size = cookie
            
            if min_cookie_idx != -1:
                used[min_cookie_idx] = True
                count += 1
        
        return count

# Test
sol = Solution()
print(sol.findContentChildren([1,2,3], [1,1]))  # 1
print(sol.findContentChildren([1,2], [1,2,3]))  # 2
````

**Complexity Analysis (Brute Force):**
- Time: O(n × m) where n is number of children and m is number of cookies
- Space: O(m) for the used array

## Approach 2: Optimal (Greedy + Two Pointers)

**Intuition:** Sort both arrays. Use a greedy approach - give the smallest cookie that satisfies each child, starting from the least greedy child. This maximizes the number of satisfied children.

**Algorithm:**
1. Sort both greed factors and cookie sizes
2. Use two pointers - one for children, one for cookies
3. If current cookie satisfies current child, move both pointers and increment count
4. If not, move to next cookie (a larger one might work)
5. Continue until we run out of children or cookies

````python
# Optimal Solution - Greedy with Two Pointers
class Solution:
    def findContentChildren(self, g: list[int], s: list[int]) -> int:
        g.sort()  # Sort greed factors
        s.sort()  # Sort cookie sizes
        
        child = 0  # Pointer for children
        cookie = 0  # Pointer for cookies
        
        while child < len(g) and cookie < len(s):
            if s[cookie] >= g[child]:
                # Cookie satisfies child
                child += 1
            # Move to next cookie regardless
            cookie += 1
        
        return child

# Test
sol = Solution()
print(sol.findContentChildren([1,2,3], [1,1]))  # 1
print(sol.findContentChildren([1,2], [1,2,3]))  # 2
print(sol.findContentChildren([10,9,8,7], [5,6,7,8]))  # 2
````

**Complexity Analysis (Optimal):**
- Time: O(n log n + m log m) for sorting, O(n + m) for traversal
- Space: O(1) extra space (in-place sorting)

## Summary
| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n × m) | O(m) |
| Greedy + Two Pointers | O(n log n + m log m) | O(1) |
