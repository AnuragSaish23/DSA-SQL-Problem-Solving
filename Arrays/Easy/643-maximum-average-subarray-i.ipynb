# Maximum Average Subarray I (LeetCode #643)

## Problem Description

You are given an integer array `nums` consisting of `n` elements, and an integer `k`.

Find a contiguous subarray whose **length is equal to** `k` that has the maximum average value and return this value. Any answer with a calculation error less than `10^-5` will be accepted.

### Example 1:
```
Input: nums = [1,12,-5,-6,50,3], k = 4
Output: 12.75000
Explanation: Maximum average is (12 + (-5) + (-6) + 50) / 4 = 51 / 4 = 12.75
```

### Example 2:
```
Input: nums = [5], k = 1
Output: 5.00000
```

### Constraints:
- `n == nums.length`
- `1 <= k <= n <= 10^5`
- `-10^4 <= nums[i] <= 10^4`

## Approach 1: Brute Force

**Intuition:** Check every subarray of size k, compute its sum, and track the maximum.

**Algorithm:**
1. For each starting index i from 0 to n-k
2. Compute the sum of elements from i to i+k-1
3. Track the maximum sum
4. Return max_sum / k

````python
# Brute Force Solution
class Solution:
    def findMaxAverage(self, nums: list[int], k: int) -> float:
        max_sum = float('-inf')
        
        for i in range(len(nums) - k + 1):
            current_sum = 0
            for j in range(i, i + k):
                current_sum += nums[j]
            max_sum = max(max_sum, current_sum)
        
        return max_sum / k
````

**Complexity Analysis (Brute Force):**
- Time: O(n × k) - For each of the n-k+1 starting positions, we sum k elements
- Space: O(1) - No extra space used

---
## Approach 2: Optimal (Sliding Window)

**Intuition:** Instead of recalculating the sum for each subarray from scratch, we can slide a window of size k across the array. When moving the window one position right, we subtract the element leaving the window and add the new element entering it.

**Algorithm:**
1. Compute the sum of the first k elements (initial window)
2. Slide the window: subtract the leftmost element, add the new rightmost element
3. Track the maximum sum
4. Return max_sum / k

````python
# Optimal Solution - Sliding Window
class Solution:
    def findMaxAverage(self, nums: list[int], k: int) -> float:
        # Compute sum of first window
        window_sum = sum(nums[:k])
        max_sum = window_sum
        
        # Slide the window
        for i in range(k, len(nums)):
            window_sum += nums[i] - nums[i - k]  # Add new element, remove old
            max_sum = max(max_sum, window_sum)
        
        return max_sum / k
````

**Complexity Analysis (Optimal):**
- Time: O(n) - Single pass through the array
- Space: O(1) - Only constant extra space

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n × k) | O(1) |
| Optimal (Sliding Window) | O(n) | O(1) |

**Key Insight:** The sliding window technique avoids redundant calculations by reusing the previous window's sum — just subtract the element going out and add the element coming in.
