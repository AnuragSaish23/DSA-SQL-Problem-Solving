# Intersection of Two Arrays II (LeetCode #350)

## Problem Description

Given two integer arrays `nums1` and `nums2`, return an array of their intersection. Each element in the result must appear as many times as it shows in both arrays and you may return the result in **any order**.

**Example 1:**
```
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2,2]
```

**Example 2:**
```
Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [4,9]
Explanation: [9,4] is also accepted.
```

**Constraints:**
- `1 <= nums1.length, nums2.length <= 1000`
- `0 <= nums1[i], nums2[i] <= 1000`

**Follow up:**
- What if the given array is already sorted? How would you optimize your algorithm?
- What if nums1's size is small compared to nums2's size? Which algorithm is better?
- What if elements of nums2 are stored on disk, and the memory is limited such that you cannot load all elements into the memory at once?

## Approach 1: Brute Force

**Intuition:** For each element in the first array, check if it exists in the second array. If found, add to result and remove from the second array to handle duplicates.

**Algorithm:**
1. Convert nums2 to a list for easy removal
2. For each element in nums1, check if it exists in the list
3. If found, add to result and remove from the list

````python
# Brute Force Solution
class Solution:
    def intersect(self, nums1: list[int], nums2: list[int]) -> list[int]:
        result = []
        nums2_list = list(nums2)
        
        for num in nums1:
            if num in nums2_list:
                result.append(num)
                nums2_list.remove(num)
        
        return result

# Test
sol = Solution()
print(sol.intersect([1,2,2,1], [2,2]))  # [2, 2]
print(sol.intersect([4,9,5], [9,4,9,8,4]))  # [4, 9] or [9, 4]
````

**Complexity Analysis (Brute Force):**
- Time: O(n × m) where n and m are lengths of nums1 and nums2
- Space: O(m) for the copy of nums2

## Approach 2: HashMap

**Intuition:** Use a hashmap to count occurrences of each element in the smaller array, then iterate through the larger array and add common elements.

**Algorithm:**
1. Count frequency of elements in nums1 using a hashmap
2. Iterate through nums2, if element exists in hashmap with count > 0, add to result and decrement count

````python
# HashMap Solution
from collections import Counter

class Solution:
    def intersect(self, nums1: list[int], nums2: list[int]) -> list[int]:
        # Use smaller array for hashmap to save space
        if len(nums1) > len(nums2):
            nums1, nums2 = nums2, nums1
        
        count = Counter(nums1)
        result = []
        
        for num in nums2:
            if count[num] > 0:
                result.append(num)
                count[num] -= 1
        
        return result

# Test
sol = Solution()
print(sol.intersect([1,2,2,1], [2,2]))  # [2, 2]
print(sol.intersect([4,9,5], [9,4,9,8,4]))  # [9, 4]
````

**Complexity Analysis (HashMap):**
- Time: O(n + m) where n and m are lengths of nums1 and nums2
- Space: O(min(n, m)) for the hashmap

## Approach 3: Optimal (Two Pointers - If Sorted)

**Intuition:** If the arrays are sorted (or can be sorted), use two pointers to find common elements efficiently.

**Algorithm:**
1. Sort both arrays
2. Use two pointers, one for each array
3. If elements are equal, add to result and move both pointers
4. If nums1[i] < nums2[j], move i forward, else move j forward

````python
# Optimal Solution - Two Pointers (for sorted arrays)
class Solution:
    def intersect(self, nums1: list[int], nums2: list[int]) -> list[int]:
        nums1.sort()
        nums2.sort()
        
        i, j = 0, 0
        result = []
        
        while i < len(nums1) and j < len(nums2):
            if nums1[i] == nums2[j]:
                result.append(nums1[i])
                i += 1
                j += 1
            elif nums1[i] < nums2[j]:
                i += 1
            else:
                j += 1
        
        return result

# Test
sol = Solution()
print(sol.intersect([1,2,2,1], [2,2]))  # [2, 2]
print(sol.intersect([4,9,5], [9,4,9,8,4]))  # [4, 9]
````

**Complexity Analysis (Two Pointers):**
- Time: O(n log n + m log m) for sorting, O(n + m) for traversal
- Space: O(1) extra space (excluding result array)

**Note:** If arrays are already sorted, this becomes O(n + m) time and O(1) space, making it optimal.

| Two Pointers | O(n log n + m log m) | O(1) |
