# Remove Duplicates from Sorted Array II (LeetCode #80)

## Problem Description

Given an integer array `nums` sorted in **non-decreasing order**, remove some duplicates **in-place** such that each unique element appears **at most twice**. The relative order of the elements should be kept the same.

Since it is impossible to change the length of the array in some languages, you must instead have the result be placed in the **first part** of the array `nums`. More formally, if there are `k` elements after removing the duplicates, then the first `k` elements of `nums` should hold the final result. It does not matter what you leave beyond the first `k` elements.

Return `k` after placing the final result in the first `k` slots of `nums`.

Do **not** allocate extra space for another array. You must do this by **modifying the input array in-place** with O(1) extra memory.

### Example 1:
```
Input: nums = [1,1,1,2,2,3]
Output: 5, nums = [1,1,2,2,3,_]
Explanation: Your function should return k = 5, with the first five elements of nums being 1, 1, 2, 2 and 3 respectively.
It does not matter what you leave beyond the returned k (hence they are underscores).
```

### Example 2:
```
Input: nums = [0,0,1,1,1,1,2,3,3]
Output: 7, nums = [0,0,1,1,2,3,3,_,_]
Explanation: Your function should return k = 7, with the first seven elements of nums being 0, 0, 1, 1, 2, 3 and 3 respectively.
```

### Constraints:
- `1 <= nums.length <= 3 * 10^4`
- `-10^4 <= nums[i] <= 10^4`
- `nums` is sorted in **non-decreasing** order.

## Approach 1: Brute Force (Using Extra Space)

**Intuition:** Use a dictionary to count occurrences and rebuild the array. This violates the O(1) space constraint but helps understand the problem.

**Algorithm:**
1. Count occurrences of each element
2. For each element, keep at most 2 occurrences
3. Rebuild the array

````python
# Brute Force Solution (uses extra space - not optimal)
from collections import Counter

class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        count = Counter(nums)
        k = 0
        
        for num in sorted(count.keys()):
            occurrences = min(count[num], 2)
            for _ in range(occurrences):
                nums[k] = num
                k += 1
        
        return k
````

**Complexity Analysis (Brute Force):**
- Time: O(n) - One pass to count, one to rebuild
- Space: O(n) - For the counter dictionary

---
## Approach 2: Optimal (Two Pointers)

**Intuition:** Use a slow pointer to track the position where the next valid element should be placed. Since array is sorted, we only need to compare with the element at position `k-2` to ensure at most 2 occurrences.

**Algorithm:**
1. If array length <= 2, return length (no duplicates to remove)
2. Initialize slow pointer `k = 2` (first two elements are always valid)
3. For each element from index 2:
   - If current element != element at position `k-2`, it's valid
   - Place it at position k and increment k
4. Return k

````python
# Optimal Solution - Two Pointers
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        if len(nums) <= 2:
            return len(nums)
        
        # k is the position where next valid element should go
        k = 2
        
        for i in range(2, len(nums)):
            # If current element is different from element at k-2
            # it means we can include this element (at most 2 occurrences)
            if nums[i] != nums[k - 2]:
                nums[k] = nums[i]
                k += 1
        
        return k
````

**Complexity Analysis (Optimal):**
- Time: O(n) - Single pass through the array
- Space: O(1) - Only using constant extra space

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force (Counter) | O(n) | O(n) |
| Optimal (Two Pointers) | O(n) | O(1) |

**Key Insight:** Since the array is sorted, comparing with the element 2 positions back ensures at most 2 occurrences of any element.
