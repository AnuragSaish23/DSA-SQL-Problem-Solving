# Longest Substring Without Repeating Characters (LeetCode #3)

## Problem Description

Given a string `s`, find the length of the **longest substring** without repeating characters.

### Example 1:
```
Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with the length of 3.
```

### Example 2:
```
Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b", with the length of 1.
```

### Example 3:
```
Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke", with the length of 3.
Notice that the answer must be a substring, "pwke" is a subsequence and not a substring.
```

### Constraints:
- `0 <= s.length <= 5 * 10^4`
- `s` consists of English letters, digits, symbols and spaces.

## Approach 1: Brute Force

**Intuition:** Check all possible substrings. For each substring, check if it has unique characters.

**Algorithm:**
1. Loop i from 0 to n-1 (start index)
2. Loop j from i to n-1 (end index)
3. Check if s[i..j] contains duplicates using a set
4. If unique, update max_len
5. If duplicates found, break inner loop (optimisation)

````python
# Brute Force Solution
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        n = len(s)
        max_len = 0
        
        for i in range(n):
            seen = set()
            for j in range(i, n):
                if s[j] in seen:
                    break
                seen.add(s[j])
                max_len = max(max_len, j - i + 1)
        
        return max_len
````

**Complexity Analysis (Brute Force):**
- Time: O(n²) - Checking all substrings
- Space: O(min(n, m)) - Set stores characters, where m is alphabet size

---
## Approach 2: Optimal (Sliding Window with Hash Map)

**Intuition:** Use a sliding window `[left, right]`. As we move `right`, if `s[right]` is already in the map and is inside the current window, we jump `left` directly to `last_seen_index + 1` to skip the duplicate. This avoids shrinking step-by-step.

**Algorithm:**
1. Initialize char_map to store index of each char
2. Initialize left = 0, max_len = 0
3. Loop right from 0 to n-1
4.   If s[right] in char_map and char_map[s[right]] >= left:
5.     Move left to char_map[s[right]] + 1
6.   Update char_map[s[right]] = right
7.   Update max_len = max(max_len, right - left + 1)
8. Return max_len

````python
# Optimal Solution - Sliding Window optimized with Hash Map
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        char_map = {}
        left = 0
        max_len = 0
        
        for right in range(len(s)):
            # If character matches one inside the current window
            if s[right] in char_map and char_map[s[right]] >= left:
                left = char_map[s[right]] + 1
            
            char_map[s[right]] = right
            max_len = max(max_len, right - left + 1)
        
        return max_len
````

**Complexity Analysis (Optimal):**
- Time: O(n) - One pass through the string
- Space: O(min(n, m)) - Hash map stores unique characters

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n²) | O(m) |
| Optimal (Sliding Window + Map) | O(n) | O(m) |

**Key Insight:** By storing the last seen index of characters, we can instantly shrink the window past the duplicate character instead of incrementing left one by one.
