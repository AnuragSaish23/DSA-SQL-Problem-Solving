# Minimum Size Subarray Sum (LeetCode #209)

## Problem Description

Given an array of positive integers `nums` and a positive integer `target`, return the **minimal length** of a subarray whose sum is greater than or equal to `target`. If there is no such subarray, return `0` instead.

### Example 1:
```
Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
Explanation: The subarray [4,3] has the minimal length under the problem constraint.
```

### Example 2:
```
Input: target = 4, nums = [1,4,4]
Output: 1
```

### Example 3:
```
Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0
```

### Constraints:
- `1 <= target <= 10^9`
- `1 <= nums.length <= 10^5`
- `1 <= nums[i] <= 10^4`

## Approach 1: Brute Force

**Intuition:** Check all possible subarrays. For each starting position `i`, expand `j` until the sum `>= target`. specific length and update min length.

**Algorithm:**
1. Initialize min_len to infinity
2. Loop i from 0 to n-1
3.   Loop j from i to n-1
4.     Calculate sum of nums[i..j]
5.     If sum >= target, update min_len = min(min_len, j - i + 1) and break (since growing larger won't help minimize length)
6. Return 0 if min_len is still infinity, else min_len

````python
# Brute Force Solution (Time Limit Exceeded)
class Solution:
    def minSubArrayLen(self, target: int, nums: list[int]) -> int:
        n = len(nums)
        min_len = float('inf')
        
        for i in range(n):
            current_sum = 0
            for j in range(i, n):
                current_sum += nums[j]
                if current_sum >= target:
                    min_len = min(min_len, j - i + 1)
                    break
        
        return 0 if min_len == float('inf') else min_len
````

**Complexity Analysis (Brute Force):**
- Time: O(n²) - Nested loops checking subarray sums
- Space: O(1) - No extra space

---
## Approach 2: Optimal (Sliding Window / Two Pointers)

**Intuition:** We uses a dynamic sliding window. Expand the window to the right by adding elements until the sum >= target. Then, try to shrink the window from the left to minimalize the length while keeping sum >= target.

**Algorithm:**
1. Initialize left = 0, current_sum = 0, min_len = infinity
2. Iterate right pointer from 0 to n-1
3.   Add nums[right] to current_sum
4.   While current_sum >= target:
5.     Update min_len = min(min_len, right - left + 1)
6.     Subtract nums[left] from current_sum and increment left
7. Return 0 if min_len is infinity, else min_len

````python
# Optimal Solution - Sliding Window
class Solution:
    def minSubArrayLen(self, target: int, nums: list[int]) -> int:
        n = len(nums)
        min_len = float('inf')
        left = 0
        current_sum = 0
        
        for right in range(n):
            current_sum += nums[right]
            
            while current_sum >= target:
                min_len = min(min_len, right - left + 1)
                current_sum -= nums[left]
                left += 1
        
        return 0 if min_len == float('inf') else min_len
````

**Complexity Analysis (Optimal):**
- Time: O(n) - Each element is added once and removed at most once
- Space: O(1) - Constant extra space

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n²) | O(1) |
| Optimal (Sliding Window) | O(n) | O(1) |

**Key Insight:** Since all numbers are positive, the sum strictly increases as we add elements and decreases as we remove them. This monotonicity allows the sliding window approach to work efficiently.
