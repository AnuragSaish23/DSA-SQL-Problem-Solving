# Boats to Save People (LeetCode #881)

## Problem Description

You are given an array `people` where `people[i]` is the weight of the `i-th` person, and an **infinite number of boats** where each boat can carry a maximum weight of `limit`. Each boat carries at most **two** people at the same time, provided the sum of the weight of those people is at most `limit`.

Return the **minimum** number of boats to carry every given person.

### Example 1:
```
Input: people = [1,2], limit = 3
Output: 1
Explanation: 1 boat (1, 2)
```

### Example 2:
```
Input: people = [3,2,2,1], limit = 3
Output: 3
Explanation: 3 boats (1, 2), (2) and (3)
```

### Example 3:
```
Input: people = [3,5,3,4], limit = 5
Output: 4
Explanation: 4 boats (3), (3), (4), (5)
```

### Constraints:
- `1 <= people.length <= 5 * 10^4`
- `1 <= people[i] <= limit <= 3 * 10^4`

## Approach 1: Brute Force

**Intuition:** Try to pair each person with the heaviest possible partner they can share a boat with. Mark paired people as visited.

**Algorithm:**
1. Sort the array
2. For each unvisited person, try to find a partner
3. Greedily try to pair lightest with heaviest
4. Count boats used

````python
# Brute Force Solution
class Solution:
    def numRescueBoats(self, people: list[int], limit: int) -> int:
        people.sort()
        visited = [False] * len(people)
        boats = 0
        
        for i in range(len(people) - 1, -1, -1):
            if visited[i]:
                continue
            visited[i] = True
            boats += 1
            
            # Try to find lightest unvisited person to pair with
            for j in range(len(people)):
                if not visited[j] and people[i] + people[j] <= limit:
                    visited[j] = True
                    break
        
        return boats
````

**Complexity Analysis (Brute Force):**
- Time: O(n²) - For each person, we search for a partner
- Space: O(n) - For the visited array

## Approach 2: Optimal (Sorting + Two Pointers)

**Intuition:** Sort the array. Use two pointers — one at the lightest person and one at the heaviest. If the lightest and heaviest can share a boat, pair them. Otherwise, the heaviest person rides alone. This greedy approach works because pairing the lightest with the heaviest maximizes the chance of sharing boats.

**Algorithm:**
1. Sort the people array
2. Initialize left = 0, right = n-1, boats = 0
3. While left <= right:
   - If people[left] + people[right] <= limit: both go on one boat, move both pointers
   - Else: heaviest person rides alone, move right pointer only
   - Increment boats
4. Return boats

````python
# Optimal Solution - Sorting + Two Pointers
class Solution:
    def numRescueBoats(self, people: list[int], limit: int) -> int:
        people.sort()
        left, right = 0, len(people) - 1
        boats = 0
        
        while left <= right:
            # If lightest and heaviest can share a boat
            if people[left] + people[right] <= limit:
                left += 1  # Lightest is paired
            
            # Heaviest person always gets on a boat (alone or paired)
            right -= 1
            boats += 1
        
        return boats
````

**Complexity Analysis (Optimal):**
- Time: O(n log n) - Sorting dominates; two-pointer scan is O(n)
- Space: O(1) - Only constant extra space (excluding sort)

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force | O(n²) | O(n) |
| Optimal (Sort + Two Pointers) | O(n log n) | O(1) |

**Key Insight:** Greedy pairing of lightest with heaviest is optimal because if the lightest can't fit with the heaviest, it can't fit with anyone heavier either.
