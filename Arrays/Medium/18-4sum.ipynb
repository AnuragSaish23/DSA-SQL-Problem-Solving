# 4Sum (LeetCode #18)

## Problem Description

Given an array `nums` of `n` integers, return an array of all the **unique quadruplets** `[nums[a], nums[b], nums[c], nums[d]]` such that:

- `0 <= a, b, c, d < n`
- `a`, `b`, `c`, and `d` are **distinct**
- `nums[a] + nums[b] + nums[c] + nums[d] == target`

You may return the answer in **any order**.

**Example 1:**
```
Input: nums = [1,0,-1,0,-2,2], target = 0
Output: [[-2,-1,1,2],[-2,0,0,2],[-1,0,0,1]]
```

**Example 2:**
```
Input: nums = [2,2,2,2,2], target = 8
Output: [[2,2,2,2]]
```

**Constraints:**
- `1 <= nums.length <= 200`
- `-10^9 <= nums[i] <= 10^9`
- `-10^9 <= target <= 10^9`

## Approach 1: Brute Force

**Intuition:** Check all possible combinations of 4 elements using 4 nested loops.

**Algorithm:**
1. Use 4 nested loops to iterate through all combinations
2. Check if sum equals target
3. Use a set to avoid duplicates

````python
# Brute Force Solution
class Solution:
    def fourSum(self, nums: list[int], target: int) -> list[list[int]]:
        n = len(nums)
        result = set()
        nums.sort()  # Sort to avoid duplicates easily
        
        for i in range(n):
            for j in range(i + 1, n):
                for k in range(j + 1, n):
                    for l in range(k + 1, n):
                        if nums[i] + nums[j] + nums[k] + nums[l] == target:
                            result.add((nums[i], nums[j], nums[k], nums[l]))
        
        return [list(quad) for quad in result]
````

**Complexity Analysis (Brute Force):**
- Time: O(n^4) - Four nested loops
- Space: O(n) - To store results (not counting input)

## Approach 2: Better - Using HashSet

**Intuition:** Fix three elements and use a hash set to find the fourth element in O(1).

**Algorithm:**
1. Sort the array
2. Use 3 nested loops for first 3 elements
3. Use a hash set to find if `target - (a + b + c)` exists

````python
# Better Solution - Using HashSet
class Solution:
    def fourSum(self, nums: list[int], target: int) -> list[list[int]]:
        n = len(nums)
        result = set()
        nums.sort()
        
        for i in range(n):
            for j in range(i + 1, n):
                seen = set()
                for k in range(j + 1, n):
                    fourth = target - (nums[i] + nums[j] + nums[k])
                    if fourth in seen:
                        result.add((nums[i], nums[j], fourth, nums[k]))
                    seen.add(nums[k])
        
        return [list(quad) for quad in result]
````

**Complexity Analysis (Better):**
- Time: O(n^3) - Three nested loops with O(1) hash lookup
- Space: O(n) - For the hash set

## Approach 3: Optimal - Four Pointers (Two Pointer Technique)

**Intuition:** Sort the array, fix two elements with nested loops, and use two pointers for the remaining two elements. This is an extension of the 3Sum approach.

**Algorithm:**
1. Sort the array
2. Use pointer `i` from 0 to n-4, skip duplicates
3. Use pointer `j` from i+1 to n-3, skip duplicates
4. Use two pointers `left` (j+1) and `right` (n-1)
5. Calculate sum and adjust pointers based on comparison with target

````python
# Optimal Solution - Four Pointers
class Solution:
    def fourSum(self, nums: list[int], target: int) -> list[list[int]]:
        n = len(nums)
        result = []
        nums.sort()
        
        for i in range(n - 3):
            # Skip duplicates for i
            if i > 0 and nums[i] == nums[i - 1]:
                continue
            
            for j in range(i + 1, n - 2):
                # Skip duplicates for j
                if j > i + 1 and nums[j] == nums[j - 1]:
                    continue
                
                left, right = j + 1, n - 1
                
                while left < right:
                    current_sum = nums[i] + nums[j] + nums[left] + nums[right]
                    
                    if current_sum == target:
                        result.append([nums[i], nums[j], nums[left], nums[right]])
                        left += 1
                        right -= 1
                        
                        # Skip duplicates for left and right
                        while left < right and nums[left] == nums[left - 1]:
                            left += 1
                        while left < right and nums[right] == nums[right + 1]:
                            right -= 1
                    
                    elif current_sum < target:
                        left += 1
                    else:
                        right -= 1
        
        return result
````

**Complexity Analysis (Optimal):**
- Time: O(n^3) - Two nested loops with O(n) two-pointer scan
- Space: O(1) - Not counting the output array

## Summary

| Approach | Time | Space |
|----------|------|-------|
| Brute Force (4 loops) | O(n^4) | O(n) |
| HashSet | O(n^3) | O(n) |
| Four Pointers | O(n^3) | O(1) |

The Four Pointers approach is optimal as it achieves O(n^3) time with O(1) space, avoiding the need for additional hash sets.
