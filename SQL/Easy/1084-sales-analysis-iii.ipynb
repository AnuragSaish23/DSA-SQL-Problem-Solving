# Sales Analysis III (LeetCode #1084)

## Problem Description

Table: `Product`

```
+--------------+---------+
| Column Name  | Type    |
+--------------+---------+
| product_id   | int     |
| product_name | varchar |
| unit_price   | int     |
+--------------+---------+
product_id is the primary key of this table.
Each row of this table indicates the name and the price of each product.
```

Table: `Sales`

```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| seller_id   | int     |
| product_id  | int     |
| buyer_id    | int     |
| sale_date   | date    |
| quantity    | int     |
| price       | int     |
+-------------+---------+
This table can have duplicate rows.
product_id is a foreign key to the Product table.
Each row of this table contains some information about one sale.
```

Write a solution to report the products that were **only** sold in the first quarter of 2019. That is, between **2019-01-01** and **2019-03-31** inclusive.

Return the result table in **any order**.

### Example 1:

**Input:**
```
Product table:
+------------+--------------+------------+
| product_id | product_name | unit_price |
+------------+--------------+------------+
| 1          | S8           | 1000       |
| 2          | G4           | 800        |
| 3          | iPhone       | 1400       |
+------------+--------------+------------+

Sales table:
+-----------+------------+----------+------------+----------+-------+
| seller_id | product_id | buyer_id | sale_date  | quantity | price |
+-----------+------------+----------+------------+----------+-------+
| 1         | 1          | 1        | 2019-01-21 | 2        | 2000  |
| 1         | 2          | 2        | 2019-02-17 | 1        | 800   |
| 2         | 2          | 3        | 2019-06-02 | 1        | 800   |
| 3         | 3          | 4        | 2019-05-13 | 2        | 2800  |
+-----------+------------+----------+------------+----------+-------+
```

**Output:**
```
+-------------+--------------+
| product_id  | product_name |
+-------------+--------------+
| 1           | S8           |
+-------------+--------------+
```

**Explanation:**
- Product 1 (S8) was only sold in January 2019.
- Product 2 (G4) was sold in February 2019 AND June 2019 — not only in Q1.
- Product 3 (iPhone) was sold in May 2019 — not in Q1 at all.

## Approach

**Intuition:** We need to find products whose ALL sales fall within Q1 2019. This means we need products where the minimum sale date is >= '2019-01-01' AND the maximum sale date is <= '2019-03-31'.

**Algorithm:**
1. Group sales by product_id
2. Filter groups where MIN(sale_date) >= '2019-01-01' AND MAX(sale_date) <= '2019-03-31'
3. Join with Product table to get product_name

````python
# SQL Solution - Using GROUP BY with HAVING
sql_query = """
SELECT p.product_id, p.product_name
FROM Product p
JOIN Sales s ON p.product_id = s.product_id
GROUP BY p.product_id, p.product_name
HAVING MIN(s.sale_date) >= '2019-01-01' 
   AND MAX(s.sale_date) <= '2019-03-31';
"""
````

## Alternative Approach (Using NOT EXISTS)

````python
# Alternative SQL Solution using subquery
sql_query_alt = """
SELECT p.product_id, p.product_name
FROM Product p
WHERE p.product_id IN (
    SELECT product_id FROM Sales 
    WHERE sale_date BETWEEN '2019-01-01' AND '2019-03-31'
)
AND p.product_id NOT IN (
    SELECT product_id FROM Sales 
    WHERE sale_date NOT BETWEEN '2019-01-01' AND '2019-03-31'
);
"""
````

## Complexity Analysis

**GROUP BY + HAVING Approach:**
- Time: O(n) for grouping and aggregation
- Space: O(k) where k is the number of unique products

**NOT EXISTS Approach:**
- Time: O(n) for each subquery scan
- Space: O(k) for storing subquery results
