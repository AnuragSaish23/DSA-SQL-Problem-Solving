# Invalid Tweets (LeetCode #1683)

## Problem Description

Table: `Tweets`

```
+----------------+---------+
| Column Name    | Type    |
+----------------+---------+
| tweet_id       | int     |
| content        | varchar |
+----------------+---------+
tweet_id is the primary key for this table.
This table contains all the tweets in a social media app.
```

Write a SQL query to find the IDs of the invalid tweets. The tweet is invalid if the number of characters used in the content of the tweet is **strictly greater than 15**.

Return the result table in **any order**.

**Example:**

Input:
```
Tweets table:
+----------+----------------------------------+
| tweet_id | content                          |
+----------+----------------------------------+
| 1        | Vote for Biden                   |
| 2        | Let us make America great again! |
+----------+----------------------------------+
```

Output:
```
+----------+
| tweet_id |
+----------+
| 2        |
+----------+
```

**Explanation:** 
- Tweet 1 has length = 14. It is a valid tweet.
- Tweet 2 has length = 32. It is invalid (length > 15).

## Approach

**Intuition:** We need to find tweets where the content length exceeds 15 characters. Use the `LENGTH()` or `CHAR_LENGTH()` function to get the string length.

**Algorithm:**
1. Select tweet_id from the Tweets table
2. Filter where LENGTH(content) > 15

**Note:** 
- `LENGTH()` returns the length in bytes
- `CHAR_LENGTH()` returns the number of characters (use for multi-byte characters)

````python
# SQL Solution - Using LENGTH
sql_query = """
SELECT tweet_id
FROM Tweets
WHERE LENGTH(content) > 15;
"""

print(sql_query)
````

## Alternative Approach

````python
# SQL Solution - Using CHAR_LENGTH (better for multi-byte characters)
sql_query_char = """
SELECT tweet_id
FROM Tweets
WHERE CHAR_LENGTH(content) > 15;
"""

print(sql_query_char)
````

## Complexity Analysis

- **Time Complexity:** O(n) where n is the number of tweets
- **Space Complexity:** O(1) - just filtering rows
