# Customers Who Never Order (LeetCode #183)

## Problem Description

Table: `Customers`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| name        | varchar |
+-------------+---------+
id is the primary key for this table.
```

Table: `Orders`
```
+-------------+------+
| Column Name | Type |
+-------------+------+
| id          | int  |
| customerId  | int  |
+-------------+------+
id is the primary key for this table.
customerId is a foreign key referencing id from Customers table.
```

Write a solution to find all customers who never order anything.

Return the result table in **any order**.

### Example

**Input:**
```
Customers table:
+----+-------+
| id | name  |
+----+-------+
| 1  | Joe   |
| 2  | Henry |
| 3  | Sam   |
| 4  | Max   |
+----+-------+

Orders table:
+----+------------+
| id | customerId |
+----+------------+
| 1  | 3          |
| 2  | 1          |
+----+------------+
```

**Output:**
```
+-----------+
| Customers |
+-----------+
| Henry     |
| Max       |
+-----------+
```

## Approach

**Using NOT IN Subquery**

Find all customer IDs that appear in the Orders table, then select customers whose IDs are NOT in that list.

````python
# SQL Solution using NOT IN
sql_query = """
SELECT name AS Customers
FROM Customers
WHERE id NOT IN (
    SELECT customerId 
    FROM Orders
);
"""

print(sql_query)
````

## Alternative Solution (Using LEFT JOIN)

Use a LEFT JOIN and filter where the Orders table has NULL values (no matching order).

````python
# SQL Solution using LEFT JOIN
sql_query_join = """
SELECT c.name AS Customers
FROM Customers c
LEFT JOIN Orders o ON c.id = o.customerId
WHERE o.id IS NULL;
"""

print(sql_query_join)
````

## Complexity Analysis

- **Time Complexity:** O(n * m) for NOT IN approach, O(n + m) for LEFT JOIN with proper indexing.
- **Space Complexity:** O(m) for storing customer IDs from Orders table.
