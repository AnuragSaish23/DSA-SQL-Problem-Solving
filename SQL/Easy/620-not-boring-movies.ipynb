# Not Boring Movies (LeetCode #620)

## Problem Description

Table: `Cinema`

```
+----------------+----------+
| Column Name    | Type     |
+----------------+----------+
| id             | int      |
| movie          | varchar  |
| description    | varchar  |
| rating         | float    |
+----------------+----------+
id is the primary key for this table.
Each row contains information about the name of a movie, its genre, and its rating.
rating is a 2 decimal places float in the range [0, 10]
```

Write a SQL query to report the movies with an odd-numbered ID and a description that is not "boring".

Return the result table ordered by `rating` in descending order.

**Example:**

Input:
```
Cinema table:
+----+------------+-------------+--------+
| id | movie      | description | rating |
+----+------------+-------------+--------+
| 1  | War        | great 3D    | 8.9    |
| 2  | Science    | fiction     | 8.5    |
| 3  | irish      | boring      | 6.2    |
| 4  | Ice song   | Fantacy     | 8.6    |
| 5  | House card | Interesting | 9.1    |
+----+------------+-------------+--------+
```

Output:
```
+----+------------+-------------+--------+
| id | movie      | description | rating |
+----+------------+-------------+--------+
| 5  | House card | Interesting | 9.1    |
| 1  | War        | great 3D    | 8.9    |
+----+------------+-------------+--------+
```

**Explanation:** 
- We have three movies with odd-numbered IDs: 1, 3, and 5.
- The movie with ID = 3 is boring so it is not included.
- Results are ordered by rating in descending order.

## Approach

**Intuition:** We need two filtering conditions:
1. Odd-numbered ID: can use `id % 2 = 1` or `MOD(id, 2) = 1`
2. Description not equal to "boring"

**Algorithm:**
1. Filter rows where id is odd (id % 2 = 1)
2. Filter rows where description != 'boring'
3. Order by rating descending

````python
# SQL Solution - Using modulo operator
sql_query = """
SELECT *
FROM Cinema
WHERE id % 2 = 1
  AND description != 'boring'
ORDER BY rating DESC;
"""

print(sql_query)
````

## Alternative Approaches

````python
# SQL Solution - Using MOD function
sql_query_mod = """
SELECT *
FROM Cinema
WHERE MOD(id, 2) = 1
  AND description <> 'boring'
ORDER BY rating DESC;
"""

# SQL Solution - Using bitwise AND (checks if last bit is 1 = odd)
sql_query_bitwise = """
SELECT *
FROM Cinema
WHERE id & 1 = 1
  AND description != 'boring'
ORDER BY rating DESC;
"""

print("Using MOD function:")
print(sql_query_mod)
print("\nUsing Bitwise AND:")
print(sql_query_bitwise)
````

## Complexity Analysis

- **Time Complexity:** O(n) where n is the number of rows in Cinema table
- **Space Complexity:** O(1) - just filtering rows
