# User Activity for the Past 30 Days I (LeetCode #1141)

## Problem Description

Table: **Activity**

```
+---------------+---------+
| Column Name   | Type    |
+---------------+---------+
| user_id       | int     |
| session_id    | int     |
| activity_date | date    |
| activity_type | enum    |
+---------------+---------+
```
This table may have duplicate rows.
The activity_type column is an ENUM (category) of type ('open_session', 'end_session', 'scroll_down', 'send_message').
The table shows the user activities for a social media website.
Note that each session belongs to exactly one user.

Write a solution to find the **daily active user count** for a period of **30 days** ending **2019-07-27** inclusively. A user was active on someday if they made at least one activity on that day.

Return the result table in **any order**.

### Example

**Input:**

Activity table:
```
+---------+------------+---------------+---------------+
| user_id | session_id | activity_date | activity_type |
+---------+------------+---------------+---------------+
| 1       | 1          | 2019-07-20    | open_session  |
| 1       | 1          | 2019-07-20    | scroll_down   |
| 1       | 1          | 2019-07-20    | end_session   |
| 2       | 4          | 2019-07-20    | open_session  |
| 2       | 4          | 2019-07-21    | send_message  |
| 2       | 4          | 2019-07-21    | end_session   |
| 3       | 2          | 2019-07-21    | open_session  |
| 3       | 2          | 2019-07-21    | send_message  |
| 3       | 2          | 2019-07-21    | end_session   |
| 4       | 3          | 2019-06-25    | open_session  |
| 4       | 3          | 2019-06-25    | end_session   |
+---------+------------+---------------+---------------+
```

**Output:**
```
+------------+--------------+
| day        | active_users |
+------------+--------------+
| 2019-07-20 | 2            |
| 2019-07-21 | 2            |
+------------+--------------+
```

**Explanation:**
- Note that we do not include the row with activity_date='2019-06-25' because it falls outside the 30-day window.
- On 2019-07-20, users 1 and 2 were active.
- On 2019-07-21, users 2 and 3 were active.

## Approach

**Steps:**
1. Filter activities within the 30-day window (2019-06-28 to 2019-07-27 inclusive)
2. Group by `activity_date`
3. Count distinct `user_id` values for each day

**Key Insight:**
- The 30-day period ending 2019-07-27 starts from 2019-06-28 (not 2019-06-27)
- We use `DATEDIFF` or date comparison to filter the correct range

````python
# SQL Solution
sql_query = """
SELECT 
    activity_date AS day,
    COUNT(DISTINCT user_id) AS active_users
FROM Activity
WHERE activity_date BETWEEN DATE_SUB('2019-07-27', INTERVAL 29 DAY) AND '2019-07-27'
GROUP BY activity_date;
"""

print(sql_query)

# Alternative using DATEDIFF
sql_query_alt = """
SELECT 
    activity_date AS day,
    COUNT(DISTINCT user_id) AS active_users
FROM Activity
WHERE DATEDIFF('2019-07-27', activity_date) < 30
  AND activity_date <= '2019-07-27'
GROUP BY activity_date;
"""

print("\nAlternative solution:")
print(sql_query_alt)
````

## Complexity Analysis

- **Time Complexity:** O(n)
  - Where n is the number of rows in the Activity table.
  - We scan through the table once, filtering and grouping.

- **Space Complexity:** O(d)
  - Where d is the number of distinct days in the 30-day period that have activity (max 30).
