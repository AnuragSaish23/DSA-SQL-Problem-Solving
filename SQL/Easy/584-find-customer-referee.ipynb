# Find Customer Referee (LeetCode #584)

## Problem Description

Table: `Customer`

```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| name        | varchar |
| referee_id  | int     |
+-------------+---------+
id is the primary key column for this table.
Each row of this table indicates the id of a customer, their name, 
and the id of the customer who referred them.
```

Find the names of the customer that are **not referred by** the customer with `id = 2`.

Return the result table in **any order**.

**Example:**

Input:
```
Customer table:
+----+------+------------+
| id | name | referee_id |
+----+------+------------+
| 1  | Will | null       |
| 2  | Jane | null       |
| 3  | Alex | 2          |
| 4  | Bill | null       |
| 5  | Zack | 1          |
| 6  | Mark | 2          |
+----+------+------------+
```

Output:
```
+------+
| name |
+------+
| Will |
| Jane |
| Bill |
| Zack |
+------+
```

## Approach

**Intuition:** We need to find customers where `referee_id` is NOT equal to 2. However, we must also consider NULL values - customers with NULL referee_id should be included (they weren't referred by anyone, including customer 2).

**Key Insight:** In SQL, comparing NULL with any value using `!=` or `<>` returns UNKNOWN (not TRUE), so `NULL != 2` would not include NULL rows. We must explicitly handle NULLs.

**Algorithm:**
1. Select customers where referee_id is not 2 OR referee_id is NULL
2. Alternatively, use COALESCE or IFNULL to handle NULLs

````python
# SQL Solution - Explicit NULL handling
sql_query = """
SELECT name
FROM Customer
WHERE referee_id != 2 OR referee_id IS NULL;
"""

print(sql_query)
````

## Alternative Approaches

````python
# SQL Solution - Using COALESCE
sql_query_coalesce = """
SELECT name
FROM Customer
WHERE COALESCE(referee_id, 0) != 2;
"""

# SQL Solution - Using IFNULL (MySQL specific)
sql_query_ifnull = """
SELECT name
FROM Customer
WHERE IFNULL(referee_id, 0) != 2;
"""

# SQL Solution - Using NOT operator with safe comparison
sql_query_not = """
SELECT name
FROM Customer
WHERE NOT referee_id <=> 2;
"""

print("COALESCE approach:")
print(sql_query_coalesce)
print("\nIFNULL approach (MySQL):")
print(sql_query_ifnull)
````

## Complexity Analysis

- **Time Complexity:** O(n) where n is the number of rows in Customer table
- **Space Complexity:** O(1) - just filtering rows
