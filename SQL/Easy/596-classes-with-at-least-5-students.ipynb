# Classes With At Least 5 Students (LeetCode #596)

## Problem Description

Table: `Courses`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| student     | varchar |
| class       | varchar |
+-------------+---------+
(student, class) is the primary key for this table.
Each row indicates the name of a student and the class they are enrolled in.
```

Write a solution to find all the classes that have **at least five students**.

Return the result table in **any order**.

### Example

**Input:**
```
Courses table:
+---------+----------+
| student | class    |
+---------+----------+
| A       | Math     |
| B       | English  |
| C       | Math     |
| D       | Biology  |
| E       | Math     |
| F       | Computer |
| G       | Math     |
| H       | Math     |
| I       | Math     |
+---------+----------+
```

**Output:**
```
+---------+
| class   |
+---------+
| Math    |
+---------+
```

**Explanation:** Math has 6 students, so it is included.

## Approach

**GROUP BY with HAVING**

1. Group all rows by class
2. Count the number of students in each class
3. Use HAVING clause to filter classes with at least 5 students

````python
# SQL Solution
sql_query = """
SELECT class
FROM Courses
GROUP BY class
HAVING COUNT(student) >= 5;
"""

print(sql_query)
````

## Complexity Analysis

- **Time Complexity:** O(n) - Single pass to group and count.
- **Space Complexity:** O(k) - Where k is the number of unique classes.
