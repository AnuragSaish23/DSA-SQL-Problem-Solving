# Duplicate Emails (LeetCode #182)

## Problem Description

Table: `Person`
```
+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| email       | varchar |
+-------------+---------+
id is the primary key for this table.
Each row contains an email, which is not null.
```

Write a solution to report all the duplicate emails. Note that it's guaranteed that the email field is not NULL.

Return the result table in **any order**.

### Example

**Input:**
```
Person table:
+----+---------+
| id | email   |
+----+---------+
| 1  | a@b.com |
| 2  | c@d.com |
| 3  | a@b.com |
+----+---------+
```

**Output:**
```
+---------+
| Email   |
+---------+
| a@b.com |
+---------+
```

## Approach

**GROUP BY with HAVING**

1. Group all rows by email
2. Use HAVING clause to filter groups where count > 1 (duplicates)

````python
# SQL Solution using GROUP BY and HAVING
sql_query = """
SELECT email AS Email
FROM Person
GROUP BY email
HAVING COUNT(*) > 1;
"""

print(sql_query)
````

## Alternative Solution (Using Self Join)

````python
# SQL Solution using Self Join
sql_query_join = """
SELECT DISTINCT p1.email AS Email
FROM Person p1
JOIN Person p2 ON p1.email = p2.email AND p1.id != p2.id;
"""

print(sql_query_join)
````

## Complexity Analysis

- **Time Complexity:** O(n) for GROUP BY approach with proper indexing.
- **Space Complexity:** O(n) for storing grouped results.
