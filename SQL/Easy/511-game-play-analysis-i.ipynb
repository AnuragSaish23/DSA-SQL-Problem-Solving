# Game Play Analysis I (LeetCode #511)

## Problem Description

Table: `Activity`

```
+--------------+---------+
| Column Name  | Type    |
+--------------+---------+
| player_id    | int     |
| device_id    | int     |
| event_date   | date    |
| games_played | int     |
+--------------+---------+
(player_id, event_date) is the primary key of this table.
This table shows the activity of players of some games.
Each row is a record of a player who logged in and played a number of games (possibly 0) 
before logging out on someday using some device.
```

Write a SQL query to report the **first login date** for each player.

Return the result table in **any order**.

**Example:**

Input:
```
Activity table:
+-----------+-----------+------------+--------------+
| player_id | device_id | event_date | games_played |
+-----------+-----------+------------+--------------+
| 1         | 2         | 2016-03-01 | 5            |
| 1         | 2         | 2016-05-02 | 6            |
| 2         | 3         | 2017-06-25 | 1            |
| 3         | 1         | 2016-03-02 | 0            |
| 3         | 4         | 2018-07-03 | 5            |
+-----------+-----------+------------+--------------+
```

Output:
```
+-----------+-------------+
| player_id | first_login |
+-----------+-------------+
| 1         | 2016-03-01  |
| 2         | 2017-06-25  |
| 3         | 2016-03-02  |
+-----------+-------------+
```

## Approach

**Intuition:** We need to find the minimum (earliest) `event_date` for each player. This is a classic GROUP BY with aggregate function problem.

**Algorithm:**
1. Group the data by `player_id`
2. For each group, find the minimum `event_date` using MIN() function
3. Alias the result column as `first_login`

````python
# SQL Solution
sql_query = """
SELECT 
    player_id,
    MIN(event_date) AS first_login
FROM Activity
GROUP BY player_id;
"""

print(sql_query)
````

## Complexity Analysis

- **Time Complexity:** O(n) where n is the number of rows in the Activity table
- **Space Complexity:** O(m) where m is the number of unique players
